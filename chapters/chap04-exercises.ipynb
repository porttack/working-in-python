The exercises below use turtle graphics, the same `jupyturtle` module from the chapter. Run the next two cells before you start -- they set up the turtle and give you the `jump` function from the chapter's exercise intro (moves the turtle without leaving a trail).

````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py')

from jupyturtle import make_turtle, forward, left, right, penup, pendown
````

````python
def jump(length):
    """Move forward length units without leaving a trail.

    Postcondition: Leaves the pen down.
    """
    penup()
    forward(length)
    pendown()
````
